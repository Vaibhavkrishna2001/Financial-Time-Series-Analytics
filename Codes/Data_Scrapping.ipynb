## Dataset Cleaning

### Identify the Bad Rows

```python
import pandas as pd

df = pd.read_csv("all_bank_stock_data.csv")
```

```python
df[df['Date'].isna()]
```

## Removing bad rows

```python
df = df.dropna(
    subset=['Date']
)
```

### Verifing

```python
df[df['Date'].isna()]
```

```python
print(df.shape)
```

## Converting date to datetime

```python
df['Date'] = pd.to_datetime(df['Date'])
```

### verify

```python
print(df.dtypes)
```

### Convert Numeric Columns

```python
numeric_cols = [
    'Open',
    'High',
    'Low',
    'Close',
    'Volume'
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col])
```

```python
print(df.dtypes)
```

### Currency Standardisation (USD → GBX)

```python
fx = yf.download("GBPUSD=X", start=df['Date'].min().strftime('%Y-%m-%d'))
fx = fx[['Close']].reset_index()
fx.columns = ['Date', 'GBPUSD']
fx['Date'] = pd.to_datetime(fx['Date'])

df = df.merge(fx, on='Date', how='left')
df['GBPUSD'] = df['GBPUSD'].ffill().bfill()

usd_mask = df['Bank'] == 'Goldman Sachs'

for col in ['Open', 'High', 'Low', 'Close']:
    gbx_col = f'{col}_GBX'
    df[gbx_col] = df[col]
    df.loc[usd_mask, gbx_col] = (df.loc[usd_mask, col] / df.loc[usd_mask, 'GBPUSD']) * 100
```

## Exploratory Data Analysis

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(14,6))

for bank in df['Bank'].unique():

    temp = df[df['Bank'] == bank].sort_values('Date')

    indexed = (temp['Close_GBX'] / temp['Close_GBX'].iloc[0]) * 100

    plt.plot(
        temp['Date'],
        indexed,
        label=bank
    )

plt.title(
    'Closing Stock Price Comparison (Indexed to 100 = 2 Jan 2015)'
)
plt.xlabel('Date')
plt.ylabel('Indexed Closing Price (Base = 100)')
plt.legend()
plt.show()
```

## Trading Volume

```python
plt.figure(figsize=(14,6))

for bank in df['Bank'].unique():

    temp = df[df['Bank'] == bank]

    plt.plot(
        temp['Date'],
        temp['Volume'],
        label=bank
    )

plt.yscale('log')
plt.title(
    'Trading Volume Comparison (log scale)'
)
plt.xlabel('Date')
plt.ylabel('Volume (log scale)')
plt.legend()
plt.show()
```

## Distribution of Closing Prices

```python
banks = df['Bank'].unique()

fig, axes = plt.subplots(1, len(banks), figsize=(15, 5), sharey=True)

colors = {'Lloyds': 'tab:blue', 'Barclays': 'tab:orange', 'Goldman Sachs': 'tab:green'}

for ax, bank in zip(axes, banks):

    temp = df[df['Bank'] == bank]

    ax.hist(
        temp['Close_GBX'],
        bins=30,
        color=colors.get(bank)
    )

    ax.set_title(bank)
    ax.set_xlabel('Closing Price (GBX / pence)')

axes[0].set_ylabel('Frequency')
fig.suptitle('Distribution of Closing Prices by Bank')
plt.tight_layout()
plt.show()
```

## Volatility Analysis

### Daily Returns

```python
df = df.sort_values(
    ['Bank','Date']
)

df['Daily_Return'] = (
    df.groupby('Bank')['Close']
      .pct_change()
)
```

### Volatility Calculation

```python
volatility = df.groupby(
    'Bank'
)['Daily_Return'].std()
print(volatility)
```

### Volatility Bar Chart

```python
volatility.plot(
    kind='bar',
    figsize=(8,5)
)
plt.title(
    'Volatility Comparison'
)
plt.ylabel(
    'Standard Deviation of Daily Returns'
)
plt.show()
```
